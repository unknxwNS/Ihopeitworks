```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import kagglehub
import os

# Download latest version
path = kagglehub.dataset_download("mahmoudshogaa/titanic-dataset")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv(os.path.join(path, "titanic.csv"))

df.head()
```

```python
df.info
```

```python
df.isnull().sum()
```

```python
# Fill missing values
df['Age'] = df['Age'].fillna(df['Age'].median(skipna=True))

df.drop(['Cabin'], axis=1, inplace=True) # вирішили видалити колонку
df.drop(['Embarked'], axis=1, inplace=True)
```

```python
df.isnull().sum()
```

```python
df = df[['Survived', 'Pclass', 'Sex', 'Age', 'Fare']]

df.head()
```

```python
df['Sex'].value_counts()
```

```python
# Convert categorical data to numerical data
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

print(f'Data types of the dataset:\n{df.dtypes}')
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.tail()
```

```python
# Calculate share of survivors by sex
male_survival_rate = df[(df['Sex'] == 0) & (df['Survived'] == 1)].shape[0] / df[(df['Sex'] == 0)].shape[0]
female_survival_rate = df[(df['Sex'] == 1) & (df['Survived'] == 1)].shape[0] / df[(df['Sex'] == 1)].shape[0]

print(f'Share of survived males {male_survival_rate * 100:.2f}%')
print(f'Share of survived females {female_survival_rate * 100:.2f}%')

print(f'Correlation between sex and survival: {df["Sex"].corr(df["Survived"])}')
```

```python
class1_surv_rate = df[(df['Pclass'] == 1) & (df['Survived'] == 1)].shape[0] / df[df['Pclass'] == 1].shape[0]
class2_surv_rate = df[(df['Pclass'] == 2) & (df['Survived'] == 1)].shape[0] / df[df['Pclass'] == 2].shape[0]
class3_surv_rate = df[(df['Pclass'] == 3) & (df['Survived'] == 1)].shape[0] / df[df['Pclass'] == 3].shape[0]

print(f'Share of survived class 1 passengers {class1_surv_rate * 100:.2f}%')
print(f'Share of survived class 2 passengers {class2_surv_rate * 100:.2f}%')
print(f'Share of survived class 3 passengers {class3_surv_rate * 100:.2f}%')
```

```python
survived_mean_age = df[df['Survived'] == 1]['Age'].mean()
not_survived_mean_age = df[df['Survived'] == 0]['Age'].mean()

print(f'Mean age of survived passengers: {survived_mean_age:.2f}')
print(f'Mean age of not survived passengers: {not_survived_mean_age:.2f}')
```

```python
# 14. Визначити середній вік тих, хто вижив, і тих, хто не вижив

survived_mean_age = df[df['Survived'] == 1]['Age'].mean()
not_survived_mean_age = df[df['Survived'] == 0]['Age'].mean()

age_survival_corr = df['Age'].corr(df['Survived'])

print(f'Середній вік тих, хто вижив: {survived_mean_age:.2f}')
print(f'Середній вік тих, хто не вижив: {not_survived_mean_age:.2f}')
print(f'Кореляція між віком і виживанням: {age_survival_corr:.3f}')

if abs(age_survival_corr) < 0.1:
    print('Вік майже не впливає на виживання.')
elif age_survival_corr > 0:
    print('Вік має слабкий позитивний вплив на виживання.')
else:
    print('Вік має слабкий негативний вплив на виживання.')
```

```python
# Розподіл пасажирів на групи за рівнями тарифів
df['FareGroup'] = pd.qcut(df['Fare'], q=4, labels=['Low', 'Medium', 'High', 'Very High'])

survival_by_fare = df.groupby('FareGroup', observed=True)['Survived'].mean()

print('Рівень виживання для кожної групи тарифів:')
print((survival_by_fare * 100).round(2))

if survival_by_fare.is_monotonic_increasing:
    print('Чим вищий тариф, тим вищі шанси на виживання.')
else:
    print('Вищий тариф загалом пов’язаний з вищим рівнем виживання.')
```

```python
# Середній тариф для кожного класу
mean_fare_by_class = df.groupby('Pclass')['Fare'].mean()

print('Середній тариф для кожного класу:')
print(mean_fare_by_class.round(2))

print('Існує значна різниця у тарифах між класами, оскільки середній тариф суттєво відрізняється для 1, 2 та 3 класів.')
```

```python
# Середній вік пасажирів для кожного класу
mean_age_by_class = df.groupby('Pclass')['Age'].mean()

print('Середній вік пасажирів для кожного класу:')
print(mean_age_by_class.round(2))

print('Розподіл віку за класами відрізняється: пасажири 1 класу в середньому старші, а пасажири 3 класу — молодші.')
```
