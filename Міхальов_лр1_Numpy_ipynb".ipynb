Додатково знайдіть сам кут у градусах.

````python
# Код студента
a = np.array([1, 2])
b = np.array([2, 1])

cos_angle = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle = np.degrees(np.arccos(cos_angle))

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)

Z = (A - mu) / sigma

print(Z)
````

Дано результати продажів трьох товарів за чотири місяці:

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
# Код студента
sales = np.array([
    [120, 150, 170, 160],
    [80, 90, 100, 110],
    [200, 180, 210, 230]
])

print("Загальний обсяг продажів кожного товару:", np.sum(sales, axis=1))
print("Загальний обсяг продажів у кожному місяці:", np.sum(sales, axis=0))
print("Товар із найбільшими сумарними продажами:", np.argmax(np.sum(sales, axis=1)) + 1)
print("Місяць із найбільшим загальним обсягом продажів:", np.argmax(np.sum(sales, axis=0)) + 1)
print("Середній обсяг продажів кожного товару:", np.mean(sales, axis=1))
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
matrix = np.random.randint(1, 21, (5, 5))

print("Матриця:")
print(matrix)

print("Сума головної діагоналі:", np.trace(matrix))
print("Сума побічної діагоналі:", np.trace(np.fliplr(matrix)))
print("Максимальний елемент кожного рядка:", np.max(matrix, axis=1))
print("Мінімальний елемент кожного стовпця:", np.min(matrix, axis=0))

mean = np.mean(matrix)
matrix[matrix < mean] = 0

print("Середнє значення всієї матриці:", mean)
print("Матриця після заміни:")
print(matrix)
````

### Завдання 38
Дано матрицю ознак трьох об’єктів:

```text
X = [[2, 4, 6],
     [1, 3, 5],
     [5, 7, 9]]
```

Обчисліть **евклідову відстань** між:
- першим і другим рядком;
- першим і третім рядком;
- другим і третім рядком.

Використайте операції NumPy та `np.linalg.norm()`.

````python
# Код студента
X = np.array([
    [2, 4, 6],
    [1, 3, 5],
    [5, 7, 9]
])

print("Відстань між першим і другим рядком:", np.linalg.norm(X[0] - X[1]))
print("Відстань між першим і третім рядком:", np.linalg.norm(X[0] - X[2]))
print("Відстань між другим і третім рядком:", np.linalg.norm(X[1] - X[2]))
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
# Код студента
A = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8],
    [2, 4, 6, 8],
    [3, 5, 7, 9]
])

norms = np.linalg.norm(A, axis=1)
normalized_A = A / norms[:, np.newaxis]

print("Норми рядків:", norms)
print("Нормалізована матриця:")
print(normalized_A)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)

print("Розв'язок системи:", x)

print("A @ x:")
print(A @ x)

print("Вектор b:")
print(b)

print("Результат A @ x дорівнює b:", np.allclose(A @ x, b))
````
